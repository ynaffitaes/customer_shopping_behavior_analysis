```python
import pandas as pd


df= pd.read_csv("customer_shopping_behavior.csv")
df.head()
```

```python
df.info()
```

```python
#standardising the headers
df.columns= df.columns.str.lower().str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)': 'purchase_amount'})
df.head()
```

```python
df.describe()
```

```python
df.isna().sum()
```

```python
#The review missing values will be inputted with the median based on the category
df['review_rating'] = df.groupby('category')['review_rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isna().sum()
```

Feature engineering

```python
#creating a column age_group
labels= ['Young Adult', 'Middle-aged', 'Senior']
df["age_group"]= pd.qcut(df["age"], q=3, labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
#creating a column purchase_frequency_days

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 months': 90
    }

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
df[['discount_applied','promo_code_used']].head(10)
```
